```python
import matplotlib.pyplot as plt
import pandas as pd



bangalore_data = data[data['City'] == 'Bangalore'].sort_values(by='Day')
delhi_data = data[data['City'] == 'Delhi'].sort_values(by='Day')

bangalore_data['Avg_Temperature_C'] = (bangalore_data['Max_Temperature_C'] + bangalore_data['Min_Temperature_C']) / 2

delhi_data['Avg_Temperature_C'] = (delhi_data['Max_Temperature_C'] + delhi_data['Min_Temperature_C']) / 2

plt.figure(figsize=(10, 6))

# Plot for Bangalore: Day on X-axis, Avg Temperature on Y-axis
plt.plot(bangalore_data['Day'], bangalore_data['Avg_Temperature_C'], marker='o', linestyle='-', label='Bangalore Avg Temp')


# Plot for Delhi: Day on X-axis, Avg Temperature on Y-axis
plt.plot(delhi_data['Day'], delhi_data['Avg_Temperature_C'], marker='^', linestyle='--', label='Delhi Avg Temp')

plt.xlabel("Day of July")
plt.ylabel("Average Temperature (C)")
plt.title("Average Temperature vs. Day of July for Bangalore and Delhi")
plt.legend()
plt.grid(True)
plt.show()
```

### Scatter Plot with Customizations

Beyond simple line plots, Matplotlib allows for rich scatter plots. Here, we'll create a scatter plot using the same `x` and `y` data, but we'll add custom markers, colors, and a different title to distinguish it.

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='red', marker='o', s=100, label='Data Points')
plt.title("Scatter Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.legend()
plt.show()
```

### Combining Plots with Subplots

For more advanced visualizations, you often need to display multiple plots together. `plt.subplot()` allows you to create a grid of plots within a single figure. The arguments are `(rows, columns, plot_number)`.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# First subplot: Line Plot
axes[0].plot(x, y, color='blue', linestyle='--', marker='o')
axes[0].set_title('Line Plot (Subplot)')
axes[0].set_xlabel('X Axis')
axes[0].set_ylabel('Y Axis')
axes[0].grid(True)

# Second subplot: Scatter Plot
axes[1].scatter(x, y, color='green', marker='x', s=150, label='Custom Points')
axes[1].set_title('Scatter Plot (Subplot)')
axes[1].set_xlabel('X Axis')
axes[1].set_ylabel('Y Axis')
axes[1].grid(True)
axes[1].legend()

plt.tight_layout() # Adjusts plot parameters for a tight layout
plt.show()
```

### Pair Plot (Scatter Matrix) using Seaborn: Exploring Multivariate Relationships

For exploring relationships between all pairs of variables in a dataset, a pair plot (or scatter matrix) is incredibly powerful. It plots scatter plots for each pair of numerical variables and histograms for single variables. We'll use the `seaborn` library for this, as it provides a higher-level interface built on Matplotlib for statistical graphics.

```python
import seaborn as sns

sns.pairplot(df, hue='target', palette='husl')
plt.suptitle('Pair Plot of Iris Dataset Features', y=1.02) # Adjust suptitle position
plt.show()
```
