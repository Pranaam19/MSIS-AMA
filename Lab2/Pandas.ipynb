```python
#Assignment
# 1.Total no. of passengers
# 2.find the avg age of survivers and non survivers
# 3.List the passengers who are older than 60 years
# 4.Find Average fare for each passeger class

df=pd.read_csv("/content/drive/MyDrive/Titanic-Dataset.csv")

# 1. Total no. of passengers
total_passengers = len(df)
print(f"Total number of passengers: {total_passengers}")

# 2. Average age of survivors and non-survivors
avg_age_by_survival = df.groupby('Survived')['Age'].mean()
print("\nAverage age by survival status:")
print(avg_age_by_survival)

# 3. List the passengers who are older than 60 years
passengers_older_than_60 = df[df['Age'] > 60]
print("\nPassengers older than 60 years:")
print(passengers_older_than_60[['Name', 'Age', 'Survived', 'Pclass']])

# 4. Find Average fare for each passenger class
avg_fare_by_pclass = df.groupby('Pclass')['Fare'].mean()
print("\nAverage fare for each passenger class:")
print(avg_fare_by_pclass)
```
