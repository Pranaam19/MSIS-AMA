```python
import numpy as np

weather1 = np.array([
  25.1, 26.1, 25.7, 25.5, 24.8, 24.8, 24.6, 26.0, 25.5, 25.7,
    24.5, 26.1, 25.9, 24.9, 24.8, 24.8, 25.0, 25.4, 25.2, 25.0,
    25.5, 24.7, 25.0, 25.1, 25.3, 25.8, 24.8, 25.4, 25.5, 24.6,
    25.5
])

# 1. Hottest and Coldest Days & Dates
max_temp = np.max(weather1)
min_temp = np.min(weather1)

hottest_dates = np.where(weather1 == max_temp)[0] + 1
coldest_dates = np.where(weather1 == min_temp)[0] + 1

print(f"Hottest day: July {hottest_dates[0]} ({max_temp}°C)")
print(f"Coldest day: July {coldest_dates[0]} ({min_temp}°C)")

# 2.Monthly Average
avg = np.mean(weather1)
print(f"Avg Temperature: {avg:.2f}°C")

# 3. Days Reported Above 35°C
days_above_35 = np.sum(weather1 >= 30)
print(f"Days above 35°C: {days_above_35}")
```

```python
import numpy as np

# 2D Array for July (31 days) formatted as: [Min_Temp, Max_Temp]
july_weather_2d = np.array([
    [23.5, 28.2], [24.0, 29.1], [23.8, 28.7], [23.6, 28.5], [23.1, 27.8],
    [23.0, 27.8], [22.8, 27.6], [24.1, 29.0], [23.7, 28.5], [23.9, 28.7],
    [22.6, 27.5], [24.2, 29.1], [24.0, 28.9], [23.2, 27.9], [23.1, 27.8],
    [23.1, 27.8], [23.3, 28.0], [23.6, 28.4], [23.5, 28.2], [23.3, 28.0],
    [23.7, 28.5], [22.9, 27.7], [23.3, 28.0], [23.4, 28.1], [23.5, 28.3],
    [23.9, 28.8], [23.1, 27.8], [23.6, 28.4], [23.7, 28.5], [22.8, 27.6],
    [23.7, 28.5]
])

# Extracting Min and Max columns using slicing
min_temps = july_weather_2d[:, 0]
max_temps = july_weather_2d[:, 1]

# 1. Hottest Day (Look at Max column) and Coldest Day
hottest_temp = np.max(max_temps)
coldest_temp = np.min(min_temps)

# Add 1 to convert 0-based index to calendar days
hottest_days = np.where(max_temps == hottest_temp)[0] + 1
coldest_days = np.where(min_temps == coldest_temp)[0] + 1

print(f"Hottest Temp: {hottest_temp}°C on July {hottest_days.tolist()}")
print(f"Coldest Temp: {coldest_temp}°C on July {coldest_days.tolist()}")

# 2. Find Monthly Averages
print(f"Monthly Average Max: {np.mean(max_temps):.2f}°C")
print(f"Monthly Average Min: {np.mean(min_temps):.2f}°C")
print(f"Overall Monthly Average: {np.mean(july_weather_2d):.2f}°C")

# 3. Days Reported Above 35°C
days_above_35 = np.sum(max_temps > 30)
print(f"Days above 30°C: {days_above_35}")
```
